### E2.32

Consider the function $f_X(x) = 1 - \frac{1}{2}x$.


```python
def fX(x):
    if 0 <= x <= 2: 
        return 1 - 0.5*x
    else:
        return 0
```

```python
# a) Plot the PDF
xs = np.linspace(-1, 4, 1000)
fXs = [fX(x) for x in xs]
sns.lineplot(x=xs, y=fXs, label="$f_X$")
```

```python
# b) Unit total axiom
from scipy.integrate import quad
quad(fX, a=0, b=2)[0]
```

```python
# c) Pr({1≤X≤2})
quad(fX, a=1, b=2)[0]
```

### E2.35

The random variable $N \sim \mathcal{N}(\mu_N=10, \sigma_N=3)$

```python
from scipy.stats import norm

muN = 10
sigmaN = 3
rvN = norm(muN, sigmaN)
```

```python
from ministats import plot_cdf
ax = plot_cdf(rvN, xlims=[0,20], rv_name="N")
ax.set_xticks(np.linspace(0, 20, 21));
ax.set_yticks(np.linspace(0, 1, 21));
```

**a)** 
We're interested in the values of $b$
where the function $F_N$
cuts the horizontal lines $0.025$ and $0.975$,
which happens approximately at $n_\ell = n_{0.025}=4$ and $n_u = n_{0.975}=16$.

```python
# b) Using the heuristic mean
[muN - 2*sigmaN, muN + 2*sigmaN]
```

```python
# Check answer we obtained in a) and b)
quad(rvN.pdf, 4, 16)[0]
```

```python
# c) Calculate the exact 95% interval
n_l = rvN.ppf(0.025)
n_u = rvN.ppf(0.975)
[n_l, n_u]
```

```python
# Check the answer to obtained in c)
quad(rvN.pdf, n_l, n_u)[0]
```

### E2.36

Compute the expected value $\mathbb{E}_U[w(U)]$,
where $w(u) = u^3$ and $U$ is the uniform random variable $\mathcal{U}(0,1)$.


```python
def w(u):
    return u**3

from scipy.stats import uniform
alpha, beta = 0, 1
rvU = uniform(alpha, beta-alpha)
rvU.expect(w)
```
