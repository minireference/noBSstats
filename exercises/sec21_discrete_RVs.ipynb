## Python functions defined in the text

```python
# define the probability mass function for the random variable D
def fD(d):
    if d in {1,2,3,4,5,6}:
        return 1/6
    else:
        return 0
```

```python
import numpy as np
import math

def fH(h):
    lam = 20
    return lam**h * np.exp(-lam) / math.factorial(h)
```

```python
def FH(b):
    return sum([fH(h) for h in range(0,b+1)])
```
