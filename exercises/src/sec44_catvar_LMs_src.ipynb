```python
# @solution
# Fit the linear model
import statsmodels.formula.api as smf
lmcu = smf.ols("score ~ 1 + C(curriculum)", data=students).fit()

# Report the t-statistics
lmcu.tvalues
```

<!-- @solution -->
The label `C(curriculum)[T.lecture]` identifies the results

so we have failed to reject $H_0$:
we haven't seen enough evidence to think the population means are different.

<!-- @solution -->
Let's now repeat the calculations of Student's two-sample $t$-test with pooled variance
from
by calling the function `scipy.stats.ttest_ind` with the option `equal_var=True`.

```python
# @studentprompt
# Extract the data for the two groups
scoresD = ...
scoresL = ...

# Run the t-test with pooled variance (option `equal_var=True`)
from scipy.stats import ttest_ind
...
```

```python
# @solution
# Extract the data for the two groups
studentsD = students[students["curriculum"]=="debate"]
studentsL = students[students["curriculum"]=="lecture"]
scoresD = studentsD["score"]
scoresL = studentsL["score"]

# Run the t-test with pooled variance (option `equal_var=True`)
from scipy.stats import ttest_ind
ttest_ind(scoresL, scoresD, equal_var=True)
```

<!-- @solution -->
We see that the $t$-statistic and the $p$-value are exactly the same.
Indeed,
the $t$-test for the slope parameter of a binary categorical predictor
is equivalent to Student's $t$-test with pooled variance.

```python
# @solution
import statsmodels.formula.api as smf

# Fit the "short" model
formula3 = "score ~ 1 + alc + weed + exrc + C(loc)"
lm3 = smf.ols(formula3, data=doctors).fit()

# Fit the "long" model
formula3work = formula3 + " + C(work)"
lm3work = smf.ols(formula3work, data=doctors).fit()
```

```python
# @studentprompt
# Run the partial F-test to compare `lm3work` vs `lm3`
...
```

```python
# @solution
# Run the partial F-test to compare `lm3work` vs `lm3`
lm3work.compare_f_test(lm3)
```

<!-- @solution -->

the $p$-value,
and the numerator degrees of freedom of the test.
The $p$-value is $0.223$
so we have failed to reject $H_0$:
including the categorical predictor `C(work)`
doesn't significantly improve the model.

```python
# @studentprompt
# c) 
# Extract the data for each group
Avalues = ...
Bvalues = ...
Cvalues = ...
Dvalues = ...

# Apply the one-way ANOVA test
from scipy.stats import f_oneway
...
```

```python
# @solution
# c) 
# Extract the data for each group
Avalues = fourgroups[fourgroups["group"]=="A"]["value"]
Bvalues = fourgroups[fourgroups["group"]=="B"]["value"]
Cvalues = fourgroups[fourgroups["group"]=="C"]["value"]
Dvalues = fourgroups[fourgroups["group"]=="D"]["value"]

# Apply the one-way ANOVA test
from scipy.stats import f_oneway
resABCD = f_oneway(Avalues, Bvalues, Cvalues, Dvalues)
resABCD.statistic, resABCD.pvalue
```
