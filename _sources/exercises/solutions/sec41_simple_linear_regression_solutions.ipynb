### E4.1

<!-- {exercise:doctors-fit-LM-score-vs-alc} -->

Use `statsmodels` to fit a linear model to the doctors dataset
([`datasets/doctors.csv`](https://noBSstats.com/datasets/doctors.csv))
based on the formula `"score 1 + alc"`.
Interpret the estimated slope parameter in your own words.

```python
import pandas as pd
doctors = pd.read_csv("datasets/doctors.csv")
doctors.head(3)
```

```python
import statsmodels.formula.api as smf
lm_alc = smf.ols("score ~ 1 + alc", data=doctors).fit()
lm_alc.params
```

The estimated model parameters are $\widehat{\beta}_{0} = 69.06$
and $\widehat{\beta}_{1} = \widehat{\beta}_{\texttt{alc}} = -1.78$.
The estimated slope parameter $\widehat{\beta}_{\texttt{alc}}$ tells us that,
according to the model,
each additional standard unit of alcohol consumed
decreases the sleep score by $1.78$ units,
on average.

### E4.2

<!-- {exercise:simple-LM-fit-from-LA-book} -->

Consider the following dataset with six observations
$[\mathbf{x},\mathbf{y}]=$ $(105, 45)$, $(113,63)$, $(125,86)$, $(137,118)$, $(141,112)$, $(153,169)$.
We're interested in the linear model $Y|x \sim \mathcal{N}(\beta_0 + \beta_1x, \; \sigma)$
for this data.

**a)** Use the formulas for $\widehat{\beta}_{0}$ and $\widehat{\beta}_{1}$
from page 
to estimate the intercept $\beta_0$ and slope $\beta_1$ model parameters.

**b)** Use the formula from page 
to estimate the model's standard deviation parameter $\sigma$.

**c)** Use `statsmodels` to fit a linear model to this dataset.
Verify your answers to **a)** and **b)**
by accessing the appropriate attributes of the fitted model.

Hint:
For part **c)**,
you can "package" the data $\mathbf{x}=\texttt{xs}$ and $\mathbf{y}=\texttt{ys}$
as a data frame object using   `df_xys=pd.DataFrame({"x":xs,"y":ys})`.



```python
# Data
xs = [105, 113, 125, 137, 141, 153]
ys = [45, 63, 86, 118, 112, 169]
```

```python
# a) 
import numpy as np
xbar = np.mean(xs)
ybar = np.mean(ys)
num = np.sum( [(x-xbar)*(y-ybar) for x,y in zip(xs,ys)] )
denom = np.sum( [(x-xbar)**2 for x in xs])
b1 = num / denom
b0 = ybar - b1*xbar
b0, b1
```

```python
# b) 
yhats = [(b0 + b1*x) for x in xs]
residuals = [y - yhat for y, yhat in zip(ys, yhats)]
SSR = np.sum( [residual**2 for residual in residuals] )
n = len(xs)
sigmahat = np.sqrt( SSR / (n-2) )
sigmahat
```

```python
# c)
import pandas as pd
df_xys = pd.DataFrame({"x": xs, "y":ys})

import statsmodels.formula.api as smf
lm_xys = smf.ols("y ~ 1 + x", data=df_xys).fit()
lm_xys.params, np.sqrt(lm_xys.scale)
```

```python
# BONUS: use scipy.stats.linregress on the same dataset
from scipy.stats import linregress
res_xy = linregress(xs, ys)
res_xy.intercept, res_xy.slope
```

The "best fit" linear model through the data is $y = -210.4 + 2.397x$.

### E4.3

<!-- {exercise:rent-price-vs-sq-ft-from-LA-book} -->

You're looking to rent an apartment.
To get an idea of how rent depends on apartment size,
you check the classifieds and find the following offers:
a 200 sq ft studio for $\$500$,
a 300 sq ft studio for $\$620$,
a 400 sq ft small apartment for $\$750$,
a 500 sq ft one-bedroom condo for $\$890$,
a 900 sq ft two-bedroom apartment for $\$1250$,
and a 1000 sq ft apartment for $\$1300$.
Find the best-fitting linear model $p(a) = \beta_0 + \beta_1 a$
for the above data,
where $p(a)$ represents the rent price for an apartment with $a$ square feet of livable area.
What is the predicted price for a 700 sq ft apartment?

```python
areas = [200, 300, 400, 500, 900, 1000]    # in square feet
prices = [500, 620, 750, 890, 1250, 1300]  # in dollars
```

```python
# Package `areas` and `prices` as a data frame
df_apts = pd.DataFrame({"area":areas, "price":prices})

# Fit the linear model
import statsmodels.formula.api as smf
lm_apts = smf.ols("price ~ 1 + area", data=df_apts).fit()
lm_apts.params
```

The "best fit" linear model for the rent price per square foot is $y = 334.5 + 1.001x$.

```python
b0 = lm_apts.params["Intercept"]
b1 = lm_apts.params["area"]

pricehat700 = b0 + b1*700
pricehat700
```

Repeat the calculations to obtain the regression result object `lm1` for the linear model $S|e \sim \mathcal{N}(\beta_0 + \beta_1 e,  \sigma)$,
which describes the students' scores $S$ as a function of their effort $e$.

**a)** Use the method `lm1.get_prediction` to predict the mean `score`
for a student that invests $10$ hours of $\texttt{effort}$,
$\mu_{\widehat{S}}(10)$.

**b)** Construct a 90% confidence interval $\ci{\mu_{S},0.9}(10)$
for the mean score $\mu_{S}(10)$ by calling the `.conf_int` method.

**c)** Use `lm1.get_prediction` to predict the `score`
of a student that invests $10$ hours of $\texttt{effort}$,
$\hat{s}(10)$.

**d)** Report a 90% prediction interval $\ci{S,0.9}(10)$ for the `score`.

Hint:
Use the code examples from Section 4.1 as a reference.



```python
# Load the students dataset
import pandas as pd
students = pd.read_csv("datasets/students.csv")

# Fit the linear model
import statsmodels.formula.api as smf
lm1 = smf.ols("score ~ 1 + effort", data=students).fit()
lm1.params
```

```python
# a)
pred10 = lm1.get_prediction({"effort":10})
pred10.predicted
```

```python
# BONUS: compute manually from the parameters
b0, b1 = lm1.params["Intercept"], lm1.params["effort"]
shat10 = b0 + b1*10
shat10
```

```python
# b)
pred10.conf_int(alpha=0.1)
```

```python
# c)
pred10 = lm1.get_prediction({"effort":10})
pred10.predicted
```
