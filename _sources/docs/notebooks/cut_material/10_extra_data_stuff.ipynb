### Random sampling alternatives

```python
import random
random.seed(22)
```

```python
# select sample of 20 from 100 individuals
pop1 = range(1,101)
print(sorted(random.sample(pop1, 20)))
```

```python
# 5 strata of 20 individuals
strata = [range(1,21) for s in range(5)]
for statum in strata:
    print(sorted(random.sample(statum, 4)))
```

```python
# choose 2 clusters out of 10
clusters = range(1,11)
print(sorted(random.sample(clusters, 2)))
```

```python
students = pd.read_csv("../datasets/students.csv")
cov_matrix = students[["effort", "score"]].cov()

stde = students["effort"].std()
stds = students["score"].std()
denom = np.array([[stde*stde, stde*stds],
                  [stds*stde, stds*stds]])

corr_matrix = cov_matrix / denom
corr_matrix  # == students[["effort", "score"]].corr()
```
