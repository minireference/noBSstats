### E3.57

<!-- {??} -->

Irregular kombucha batch from a population $K \sim \mathcal{N}(1003,10)$.

Can we detect via $t$-test on a random sample of size $n=20$?


```python
# a) Calculate power using simulation
from scipy.stats import norm

rvK = norm(1003, 10)


np.random.seed(47)


from scipy.stats import ttest_1samp

n = 20
N = 10000
number_rejected = 0
for j in range(N):
    ksample = rvK.rvs(n)
    res = ttest_1samp(ksample, popmean=1000, alternative="two-sided")
    if res.pvalue <= 0.05:
        number_rejected = number_rejected + 1

power = number_rejected / N
power
```

```python
# b) Calculate power using `TTestPower` from `statsmodels`
Delta = 1003 - 1000
d = Delta / 10  # = effect size in Cohen's d

from statsmodels.stats.power import TTestPower
ttp = TTestPower()
ttp.power(effect_size=d, nobs=n, alpha=0.05, alternative="two-sided")
```

```python
from statsmodels.stats.power import TTestPower
ttp = TTestPower()
ttp.solve_power(effect_size=d, power=0.8, alpha=0.05, alternative="two-sided")
```
