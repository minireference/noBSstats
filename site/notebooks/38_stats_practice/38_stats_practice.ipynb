## Avoiding statistical misconceptions

### The correct interpretation of hypothesis test results

- Calculating the $p$-value
- Decision rule based on a cutoff value
- Reporting effect size estimates

### Misconceptions about $p$-values

- Probability of hypotheses
- Viewing results as absolute proof
- Probability that we made a mistake
- Probability of replication
- Interpreting the $p$-value as an effect size
- Misconceptions about confidence intervals

### Problems with hypothesis testing in general

- Sneaky language
- Rigid procedure
- A reasonable null hypothesis might not exist
- The alternative hypothesis is not tested
- Mistaking correlation for causation
- Sample size issues
- Post-hoc power calculations
