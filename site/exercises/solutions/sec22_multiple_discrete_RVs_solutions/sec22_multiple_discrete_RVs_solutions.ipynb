```python

def fV(v):
    if v == 1:
        return 0.03
    elif v == 0:
        return 0.97

def fTgivenV1(t):
    if t == 1:
        return 0.90
    elif t == 0:
        return 0.10

def fTgivenV0(t):
    if t == 1:
        return 0.20
    elif t == 0:
        return 0.80

def fTgivenV(t,v):
    if v == 1:
        return fTgivenV1(t)
    elif v == 0:
        return fTgivenV0(t)

def fTV(t,v):
    return fTgivenV(t,v) * fV(v)

for v in range(0,1+1):
    for t in range(0,1+1):
        print("Pr({T=" + str(t) + ",V=" + str(v)+"})",
              "=", fTgivenV(t,v), "*", fV(v),
              "=", round(fTV(t,v),5))

def fVgivenT(v,t):
    num = fTgivenV(t,v) * fV(v)
    denom = fTgivenV(t,0) * fV(0) + fTgivenV(t,1) * fV(1)
    return num / denom

fVgivenT(1,0)
```
